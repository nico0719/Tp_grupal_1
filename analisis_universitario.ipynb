# Análisis de Datos Universitarios
Dataset: `university_student_data.csv`  
Período: 2015–2024 | Términos: Spring y Fall

## 1. Carga y exploración inicial del dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("university_student_data.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## 2. Descripción de las columnas

| Columna | Descripción |
|---|---|
| **Year** | Año académico (2015–2024). |
| **Term** | Período del año: `Spring` (primer semestre) o `Fall` (segundo semestre). |
| **Applications** | Cantidad de solicitudes de ingreso recibidas en ese año y período. |
| **Admitted** | Cantidad de postulantes admitidos (subconjunto de Applications). |
| **Enrolled** | Cantidad de admitidos que efectivamente se inscribieron. |
| **Retention Rate (%)** | Porcentaje de estudiantes que continúan sus estudios al siguiente período. |
| **Student Satisfaction (%)** | Porcentaje de satisfacción reportado por los estudiantes en encuestas. |
| **Engineering Enrolled** | Estudiantes inscriptos en la carrera de Ingeniería. |
| **Business Enrolled** | Estudiantes inscriptos en la carrera de Negocios/Administración. |
| **Arts Enrolled** | Estudiantes inscriptos en la carrera de Artes. |
| **Science Enrolled** | Estudiantes inscriptos en la carrera de Ciencias. |

## 3. Tasa de retención promedio por año

```python
retencion_anual = df.groupby("Year")["Retention Rate (%)"].mean()

fig, ax = plt.subplots(figsize=(10, 5))
retencion_anual.plot(ax=ax, marker="o", color="steelblue")
ax.set_title("Tasa de retención promedio por año")
ax.set_xlabel("Año")
ax.set_ylabel("Retención (%)")
ax.grid(True)
plt.tight_layout()
plt.show()
```

## 4. Satisfacción estudiantil por año

```python
satisfaccion_anual = df.groupby("Year")["Student Satisfaction (%)"].mean().reset_index()

fig, ax = plt.subplots(figsize=(10, 5))
sns.lineplot(data=satisfaccion_anual, x="Year", y="Student Satisfaction (%)", marker="o", color="coral", ax=ax)
ax.set_title("Satisfacción estudiantil por año")
ax.set_xlabel("Año")
ax.set_ylabel("Satisfacción (%)")
ax.grid(True)
plt.tight_layout()
plt.show()
```

## 5. Comparación Spring vs Fall — Tasa de retención por año y período

Se usa el dataframe **completo** (`df`) con `groupby(["Year", "Term"])` para ver ambos períodos simultáneamente.

```python
analisis_term = df.groupby(["Year", "Term"])["Retention Rate (%)"].mean().unstack()

fig, ax = plt.subplots(figsize=(12, 5))
analisis_term.plot(kind="bar", ax=ax, color=["#4c72b0", "#dd8452"])
ax.set_title("Retención por período: Spring vs Fall")
ax.set_xlabel("Año")
ax.set_ylabel("Retención (%)")
ax.legend(title="Período", loc="best")
ax.set_xticklabels(analisis_term.index, rotation=45)
ax.grid(True, axis="y")
plt.tight_layout()
plt.show()
```
